# Matplotlib ハンズオン

講義資料の流れに合わせて、順番にセルを埋めていく。詰まったら `final-plot.ipynb` を参照する。

```python
# ライブラリのインポート
import matplotlib.pyplot as plt
import matplotlib_fontja
import pandas as pd
```

```python
# TSV ファイルを読み込む
# ヒント: pandas の read_csv と sep="\t" を使う
df = pd.read_csv("./data/voltage.tsv", sep="\t")
df
```

```python
# 2 本の折れ線グラフを描く
# time を横軸、V(n002) と I(L1) を縦軸にする
plt.plot(df["time"], df["V(n002)"], label="V(n002)")
plt.plot(df["time"], df["I(L1)"], label="I(L1)")

# 軸ラベル、タイトル、凡例を追加する
plt.xlabel("時間 (s)")
plt.ylabel("値")
plt.title("電流電圧特性")
plt.legend()
plt.grid(True)

plt.show()
```
