```python
#ライブラリのインポート
import matplotlib.pyplot as plt
import matplotlib_fontja
import numpy as np
import pandas as pd
```

```python
# CSV/TSVファイルの読み込み
# 今回は Tab区切りのTSVファイルを読み込むため sep="\t" を指定
df = pd.read_csv("./data/voltage.tsv", sep="\t")
df
```

```python
plt.plot(df["time"], df["V(n002)"])
plt.plot(df["time"], df["I(L1)"])
plt.xlabel("時間 (s)")
plt.ylabel("電圧 (V)")
plt.title("電流電圧特性")

plt.show()
```
