# Matplotlib 完成版

`matplotlib/data/voltage.tsv` を読み込み、電圧と電流を同じグラフに描画する。

```python
# ライブラリのインポート
import matplotlib.pyplot as plt
import matplotlib_fontja
import pandas as pd
```

```python
# TSV ファイルを読み込む
df = pd.read_csv("./data/voltage.tsv", sep="\t")
df.head()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df["time"], df["V(n002)"], label="V(n002)", linewidth=2)
plt.plot(df["time"], df["I(L1)"], label="I(L1)", linewidth=2)
plt.xlabel("時間 (s)")
plt.ylabel("値")
plt.title("電流電圧特性")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
